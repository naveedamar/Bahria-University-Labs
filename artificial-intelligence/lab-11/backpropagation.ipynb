```python
import numpy as np

# Sigmoid
def sigmoid(x): return 1 / (1 + np.exp(-x))
def sigmoid_derivative(x): return x * (1 - x)

X = np.array([[0,0], [0,1], [1,0], [1,1]])
# Target outputs
y = np.array([[0], [1], [1], [0]])


weights_input_hidden = np.random.uniform(size=(2, 2))
weights_hidden_output = np.random.uniform(size=(2, 1))

# Training loop
for epoch in range(10000):
    # forward
    hidden_layer = sigmoid(np.dot(X, weights_input_hidden))
    output = sigmoid(np.dot(hidden_layer, weights_hidden_output))

    error = y - output

    # backpropagation
    d_output = error * sigmoid_derivative(output)
    error_hidden = d_output.dot(weights_hidden_output.T)
    d_hidden = error_hidden * sigmoid_derivative(hidden_layer)

    # update
    weights_hidden_output += hidden_layer.T.dot(d_output) * 0.1
    weights_input_hidden += X.T.dot(d_hidden) * 0.1

print("Final Output after training:\n", output)
```
