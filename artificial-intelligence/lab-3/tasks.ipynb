```python
#task 1
number = int(input("Enter a number: "))
for num in range(1, number):
    if num % 3 == 0:
        print("Boom !")
    elif num % 5 == 0:
        print("Bang !")
    elif num % 3 == 0 and num % 5 == 0:
        print("BoomBang !")
    else:
        print(num)
```

```python
#task 2
data = {3, 4, 70, 560, 1000, 1001, 303}
even = 0
odd = 0
for num in data:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even", even)
print("Odd", odd)

if even > odd:
    print("Data has more even numbers")
else:
    print("Data has even numbers")
```
