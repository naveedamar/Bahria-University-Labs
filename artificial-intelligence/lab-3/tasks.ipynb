```python
#task 1
def calculate_sum(n):
    if n <= 1:
        return n
    else:
        return n + calculate_sum(n - 1)


num = int(input("Enter a positive integer n: "))

if num < 0:
    print("Please enter a positive integer.")
else:
    result = calculate_sum(num)
    print(f"The sum of numbers from 1 to {num} is: {result}")   
```

```python
#task 2
class Seat:
    def __init__(self, seat_number, seat_type):
        self.seat_number = seat_number
        self.seat_type = seat_type
        self.base_price = 500

    def calculate_ticket_price(self):
        return self.base_price

class RegularSeat(Seat):
    pass

class VIPSeat(Seat):
    def calculate_ticket_price(self):
        return self.base_price + 500

print("--- Movie Theatre Booking ---")
print("1. Regular Seat (Rs. 500)")
print("2. VIP Seat (Rs. 1000)")

choice = input("Enter choice (1 or 2): ")
s_num = input("Enter seat number (e.g., A10): ")

if choice == "1":
    booking = RegularSeat(s_num, "Regular")
elif choice == "2":
    booking = VIPSeat(s_num, "VIP")
else:
    booking = None
    print("Invalid choice")

if booking:
    price = booking.calculate_ticket_price()
    print(f"\nBooking Confirmed!")
    print(f"Seat No: {booking.seat_number} | Type: {booking.seat_type}")
    print(f"Total Price: Rs. {price}")
```
