```python
#Factorial
def fact(n):
    if n == 0:
        return 1
    else:
        print(f"{n} x {n-1} = {n*(n-1)}")
        return n * fact(n-1)


print(f"+\n\t   {fact(5)}")
```

```python
#OOP

class Laptop:

    """
    Laptop class that represents a laptop
    """
    def __init__(self, name, model, price_range):
        self.name = name
        self.model =  model
        self.price_range = price_range

    def specification(self):
        print("Laptop Specification")
        print("Name: ", self.name)
        print("Model: ", self.model)
        print("Price Range: ", self.price_range)

mylaptop = Laptop("Lenovo", "Slim-3", "$-$$$")
mylaptop.specification()
```

```python
#make child class
class MyLaptop(Laptop):
    """
    MyLaptop class represents my laptop
    """
    def __init__(self, processor, ram):
        self.processor = processor
        self.ram = ram
    def specifications(self):
        mylaptop.specification()
        print("\nMy Laptop Specification")
        print("Processor: ", self.processor)
        print("RAM: ", self.ram)

mylap = MyLaptop("i5", 16)
mylap.specifications()
```

```python
#task 1: Fibonacci Series
def fib(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fib(n-1) + fib(n-2)

print(fib(4))
```

```python
#task 2: booking system
class Seat:
    """
    Seat class that defins seats
    """
    def __init__(self, seat_no, seat_type):
        self.seat_no = seat_no
        self.seat_type = seat_type

    def calculate_ticket_price(self):
        if self.seat_type == "regular":
            return 100
        elif self.seat_type == "vip":
            return 500

class Booking(Seat):
    """
    book class, Sub class of Seats
    """
    def __init__(self, seat_no, seat_type):
        super().__init__(seat_no, seat_type)

print("Welcome to booking seats")

seat_no = int(input("Enter your seat number: "))
seat_type = (input("Enter your seat type")).lower()

booking = Booking(seat_no, seat_type)
booking.calculate_ticket_price()
```
