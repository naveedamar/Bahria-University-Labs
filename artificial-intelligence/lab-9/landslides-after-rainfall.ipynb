```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('catalog.csv') # Replace with your filename

# normalizing
df['landslide_size'] = df['landslide_size'].str.capitalize()
plt.figure(figsize=(10,6))
top_countries = df['country_name'].value_counts().head(10)
sns.barplot(x=top_countries.values, y=top_countries.index)
plt.title('Top 10 Countries by Landslide Events')
plt.show()
```
