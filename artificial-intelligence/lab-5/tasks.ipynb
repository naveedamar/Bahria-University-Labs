```python
import heapq

graph = {
    'S': [('A', 6), ('B', 2), ('C', 1)],
    'A': [('D', 3), ('G', 20)],
    'B': [('E', 6), ('D', 2)],
    'C': [('B', 3), ('F', 4), ('E', 6)],
    'D': [('F', 5)],
    'E': [('G', 2)],
    'F': [('G', 1)]
}

heuristics = {
    'S': 6,
    'A': 8,
    'B': 6,
    'C': 5,
    'D': 4,
    'E': 2,
    'F': 1,
    'G': 0
}


def a_star(graph, start, goal):

    priority_queue = [(heuristics[start], start, [start], 0)]
    visited = set()

    while priority_queue:
        f, current_node, path, g = heapq.heappop(priority_queue)

        if current_node in visited:
            continue

        if current_node == goal:
            return path, g

        visited.add(current_node)

        for neighbor, weight in graph.get(current_node, []):
            if neighbor not in visited:
                new_g = g + weight
                new_f = new_g + heuristics.get(neighbor, 0)
                new_path = path + [neighbor]
                heapq.heappush(priority_queue, (new_f, neighbor, new_path, new_g))

    return None, float('inf')


path, total_cost = a_star(graph, 'S', 'G')

if path:
    print(f"Path found: {' -> '.join(path)}")
    print(f"Total Cost: {total_cost}")
else:
    print("No path found.")
```
