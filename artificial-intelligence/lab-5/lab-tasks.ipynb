```python
"""
Using GA, it is required to find the maximum of a function, f(x) = (x + 3y*5z *9w)/35 where x, y, z,
and w are genes to form a chromosome. Write a python program to implement and show the best individual along with best fitness,
average fitness and poorest fitness against each generation.

For demonstration purposes, the following sample set of parameters is to be used to design the algorithm:

Number of generations: 35
The population size: 10
The mutation rate: 0.1
Number of bits in each chromosome: 4
"""
```

```python
import random

POP_SIZE = 10
GENS = 35
MUTATION_RATE = 0.1
GENES_PER_CHROMO = 4

def fitness_function(chromosome):

    x, y, z, w = chromosome
    return (x + (3*y * 5*z * 9*w)) / 35

def create_individual():
    return [random.randint(0, 1) for _ in range(GENES_PER_CHROMO)]

def mutate(chromosome):
    for i in range(len(chromosome)):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]
    return chromosome

def crossover(parent1, parent2):
    point = random.randint(1, GENES_PER_CHROMO - 1)
    child1 = parent1[:point] + parent2[point:]
    child2 = parent2[:point] + parent1[point:]
    return child1, child2

population = [create_individual() for _ in range(POP_SIZE)]

print(f"{'Gen':<5} | {'Best Fitness':<12} | {'Avg Fitness':<12} | {'Poor Fitness':<12} | {'Best Individual'}")
print("-" * 75)

for gen in range(1, GENS + 1):
    fitness_scores = [fitness_function(ind) for ind in population]

    best_fit = max(fitness_scores)
    poor_fit = min(fitness_scores)
    avg_fit = sum(fitness_scores) / POP_SIZE
    best_ind = population[fitness_scores.index(best_fit)]

    print(f"{gen:<5} | {best_fit:<12.4f} | {avg_fit:<12.4f} | {poor_fit:<12.4f} | {best_ind}")

    new_population = []
    for _ in range(POP_SIZE // 2):
        parents = random.sample(population, 4)
        parents.sort(key=lambda ind: fitness_function(ind), reverse=True)
        p1, p2 = parents[0], parents[1]

        c1, c2 = crossover(p1, p2)

        new_population.append(mutate(c1))
        new_population.append(mutate(c2))

    population = new_population
```
