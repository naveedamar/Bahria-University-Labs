```python
# Movie Ticket Booking System

age = int(input("Enter customer age: "))
day = input("Enter day of the week: ").strip().capitalize()
num_tickets = int(input("Enter number of tickets to purchase: "))
is_student = input("Are you a student? (Yes/No): ").strip().lower()

weekdays = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]
weekends = ["Saturday", "Sunday"]

if day in weekends:
    base_price = 350
else:
    base_price = 300

total_cost = base_price * num_tickets

if day in weekends:
    total_cost += total_cost * 0.15

if is_student == "yes":
    total_cost -= total_cost * 0.10

if num_tickets >= 3:
    total_cost -= total_cost * 0.05

print("\n" + "=" * 30)
print("   TICKET BOOKING SUMMARY   ")
print("=" * 30)
print(f"Day:               {day}")
print(f"Tickets:           {num_tickets}")
print(f"Base Price/Ticket: Rs. {base_price}")
print("-" * 30)
print(f"TOTAL AMOUNT:      Rs. {total_cost:.2f}")
print("=" * 30)
```
