```python
#BFS
from collections import deque

#simple friendship network
graph = {
    'You': ["Alice", "Bob"],
    'Alice': ["You", "David", "Bob"],
    'Bob': ["You", "Eve"],
    'Charlie': ["Alice"],
    'Eve': ["Bob"],
}


def bfs(graph, start_node):
    visited = set([start_node])
    print(f"visited: {visited}")
    print(f"Type of graph: {type(graph)}")

    queue = deque([start_node])

    print(f"--starting BFS at {start_node}")
    order = list()

    while queue:
        print(f"Current queue: {list(queue)}")
        node = queue.popleft()
        print(f"Visited node: {node}")
        print(f"Graph: {graph}")

        order.append(node)
        print(f"see here {type(graph)}")
        for neighbor in graph[node]:
            if neighbor not in visited:
                print(f"Found neighbor {neighbor}, Adding it to queue")
                visited.add(node)
            else:
                print(f"neighbor {neighbor}, Removing it from queue")
    return order
```

```python
print(bfs(graph, "You"))
```

```python
#DFS
graph = {
    'You': ["Alice", "Bob"],
    'Alice': ["You", "David", "Bob"],
    'Bob': ["You", "Eve"],
    'Charlie': ["Alice"],
    'Eve': ["Bob"],
}

def dfs(graph, node, visited=None):
    if visited is None:
        visited = set()

    if node not in visited:
        print(f"Visited: {node}")
        visited.add(node)
        for neighbor in graph.get(node, []):
            dfs(graph, neighbor, visited)
    return visited

print("Starting DFS:")
dfs(graph, "You")
```

```python
#task 1: DFS to find values of key inputs
ages = {
    "Naveed": 38, "Hadi": 35, "Ali": 40, "Ahmed": 55, "Mohit": "NULL"
}

graph = {
    "Naveed": ["Hadi", "Ali"],
    "Hadi": ["Ahmed"],
    "Ali": ["Mohit"],
    "Ahmed": [],
    "Mohit": []
}

visited = set()

def dfs_graph(node, target, level=0):
    if node not in visited:
        visited.add(node)

        if node == target:
            return ages[node], level

        for neighbor in graph.get(node, []):
            result = dfs_graph(neighbor, target, level + 1)
            if result:
                return result
    return None

search_key = input("Enter name to search: ")
found = dfs_graph("Naveed", search_key)

if found:
    val, lvl = found
    print(f"Found {search_key}. Value: {val}, Depth Level: {lvl}")
else:
    print("Not found in the network.")
```

```python
#task 2: Finding unique friends of friends
from collections import deque

connection_graph = {
    'You': ["Alice", "Bob"],
    'Alice': ["You", "David", "Bob"],
    'Bob': ["You", "Eve"],
    'Charlie': ["Alice"],
    'Eve': ["Bob"],
}

def dfs_user(graph, node, visited=None):
    if visited is None:
        visited = set()

    if node not in visited:
        print(f"Visiting: {node}")
        visited.add(node)
        for neighbor in graph.get(node, []):
            dfs_user(graph, neighbor, visited)
    return visited

dfs_user(connection_graph, "You")
```
